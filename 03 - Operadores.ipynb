# Realizando operações com variáveis

```python
x = input("Digite o primeiro valor:")
```

```python
y = input("Digite o segundo valor:")
```

## Conversões

```python
x = int(x)
y = int(y)
```

## Operações

```python
print("Soma: ", x + y)
print("Subtração: ", x - y)
print("Multiplicação: ", x * y)
print("Divisão: ", x / y)
```
